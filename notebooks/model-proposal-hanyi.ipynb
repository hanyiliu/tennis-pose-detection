# Model Proposal - Hanyi

The goal of this model is:
- Given input of image of back-view of tennis player playing tennis,
- Output a classification between four classes: backhand, forehand, ready position, serve.

Given the features of our dataset containing the player's skeletal pose, we therefore propose a two-stage model:
1. A pose detection model, where given input of the image, output the user's pose.
2. A pose classification model, where given input of the pose, output the classification.

## 1. Pose Detection

For pose detection, a standard architecture is:
1. Classification model which first determines the bounding box of the player (global-stage).
2. With the player's bounding box, we run another pose detection model to get the key points (local-stage).

The global stage can be done fairly easily, we can build on top of pre-trained models like YOLO, or train from scratch considering that we have 2000 samples in our dataset and clear bounding box annotations for each sample.

The local stage will be slightly more complicated. There are standard existing architetures for pose detection, such as MMPose, of which we can use for inspiration. MMPose follows a standard backbone -> neck -> head architecture, like YOLO.

At the end of this stage, we would have outputted a vector of length 18 (number of keypoints) for each sample, which represents the (x, y) coordinates of each keypoint and each keypoint's visibility.

## 2. Pose Classification

At this stage, we can use a simple neural network which takes in the keypoints vector and outputs a class prediction. This can be a simple 2 or 3 layer neural net to begin off, and we can elaborate on its size if necessary. Its output will be a softmax vector of size 4, corresponding to the 4 classes.
