CNN:

Because we have consistent and balanced classes, CNN can learn how the player looks:
    Arm extension direction
    Racket location
    Torso orientation
    Stance width
    Related limb position
CNN learns directly from pixels, without specifying geometries.
Pros:
    Used full visual information
    No dependency on keypoint quality
    Strong baseline for image task (Standard approach for pose/action recognition)
    Automatically learns features
Cons:
    Computationally expensive
    Hard to interpret (learned features not human-readable)
    Needs more data to generalize
    May overfit
    (May learn non-pose cues, racket color/background)

Keypoint-based MLP:

Uses 18 keypoints x (x,y)
Learns body geometry

Pros:
    Lightweight and fast
    Highly pose-focused
    More interpretable
    Less prone to background bias

Cons:
    Relies entirely on keypoint accuracy
    Loses visual context (cannot see racket, ball, motion, etc.)
    Limited expressiveness (cannot capture fine visual distinctions)

Works well because we have fixed 18 keypoints
High visibility across all poses
Consistent annotation format

May struggle to distinguish similar poses when poses may overlap geometrically.
