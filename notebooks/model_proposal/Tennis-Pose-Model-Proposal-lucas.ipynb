# Potential Model Architectures

## Approach A: The Lightweight "Geometry" Model (MLP)

Input: Flattened 1D array of 36 values (18 joints * 2 coordinates)

Architecture: 3-layer Deep Neural Network

Pros: Extremely fast; can run on a mobile device or a low-end CPU

## Approach B: Graph Convolutional Networks (ST-GCN)

Concept: Instead of a flat list, the model treats joints as nodes in a graph

Pros: It understands that the "elbow" is physically connected to the "wrist," making it more resilient to noisy data

## Approach C: The Vision-Pose Hybrid (CNN-RNN)

Concept: Extract features from the image using a ResNet backbone and combine them with the keypoint data

Pros: Captures external context, such as the position of the tennis racket or the net.
